```python
# ================================
# VOIS AICTE MAJOR PROJECT
# Seasonal Agriculture Performance Analysis
# ================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

plt.style.use("ggplot")

# -------------------------------
# Load Dataset
# -------------------------------
df = pd.read_csv("seasonal_agriculture_performance_dataset.csv")

print("="*60)
print("DATASET OVERVIEW")
print("="*60)
print("Shape :", df.shape)
print("\nFirst 5 Rows:")
print(df.head())

print("\nDataset Info")
print(df.info())

print("\nMissing Values")
print(df.isnull().sum())

print("\nRemoving Duplicates...")
before = df.shape[0]
df = df.drop_duplicates()
after = df.shape[0]
print(f"Removed {before-after} duplicate rows")

print("\nSummary Statistics")
print(df.describe())

# -------------------------------
# Seasonal Yield
# -------------------------------
yield_season = df.groupby("Season")["Yield_Tonnes_Ha"].mean().sort_values()

plt.figure(figsize=(6,4))
yield_season.plot(kind="bar")
plt.title("Average Yield by Season")
plt.ylabel("Tonnes/Ha")
plt.show()

# -------------------------------
# Seasonal Profit
# -------------------------------
profit = df.groupby("Season")["Profit_INR"].mean().sort_values()

plt.figure(figsize=(6,4))
profit.plot(kind="bar")
plt.title("Average Profit by Season")
plt.ylabel("Profit (INR)")
plt.show()

# -------------------------------
# Revenue vs Cost
# -------------------------------
finance = df.groupby("Season")[["Revenue_INR","Total_Cost_INR"]].mean()

plt.figure(figsize=(7,4))
finance.plot(marker="o")
plt.title("Average Revenue vs Cost")
plt.ylabel("INR")
plt.grid(True)
plt.show()

# -------------------------------
# Rainfall
# -------------------------------
rain = df.groupby("Season")["Rainfall_mm"].mean()

plt.figure(figsize=(6,4))
rain.plot(kind="bar")
plt.title("Average Rainfall")
plt.ylabel("mm")
plt.show()

# -------------------------------
# Temperature
# -------------------------------
temp = df.groupby("Season")["Avg_Temperature_C"].mean()

plt.figure(figsize=(6,4))
temp.plot(kind="bar")
plt.title("Average Temperature")
plt.ylabel("°C")
plt.show()

# -------------------------------
# Water Usage
# -------------------------------
water = df.groupby("Season")["Water_Used_m3"].mean()

plt.figure(figsize=(6,4))
water.plot(kind="bar")
plt.title("Average Water Usage")
plt.ylabel("m³")
plt.show()

# -------------------------------
# Water Efficiency
# -------------------------------
eff = df.groupby("Season")["Water_Efficiency_t_per_1000m3"].mean()

plt.figure(figsize=(6,4))
eff.plot(kind="bar")
plt.title("Water Efficiency by Season")
plt.ylabel("Efficiency")
plt.show()

# -------------------------------
# Disease Risk
# -------------------------------
risk = df.groupby("Season")["Disease_Pest_Risk_pct"].mean()

plt.figure(figsize=(6,4))
risk.plot(kind="bar")
plt.title("Disease & Pest Risk")
plt.ylabel("%")
plt.show()

# -------------------------------
# Crop Production
# -------------------------------
crop = df.groupby("Crop")["Production_Tonnes"].sum().sort_values(ascending=False)

plt.figure(figsize=(8,4))
crop.plot(kind="bar")
plt.title("Total Production by Crop")
plt.ylabel("Tonnes")
plt.show()

# -------------------------------
# State-wise Profit
# -------------------------------
state = df.groupby("State")["Profit_INR"].mean().sort_values(ascending=False)

plt.figure(figsize=(8,4))
state.plot(kind="bar")
plt.title("Average Profit by State")
plt.ylabel("INR")
plt.show()

# -------------------------------
# Irrigation Method
# -------------------------------
irr = df.groupby("Irrigation_Method")["Yield_Tonnes_Ha"].mean().sort_values()

plt.figure(figsize=(7,4))
irr.plot(kind="bar")
plt.title("Yield by Irrigation Method")
plt.ylabel("Tonnes/Ha")
plt.show()

# -------------------------------
# Correlation Matrix
# -------------------------------
corr = df.select_dtypes(include=np.number).corr()

plt.figure(figsize=(10,8))
plt.imshow(corr, cmap="coolwarm")
plt.colorbar()

plt.xticks(range(len(corr.columns)), corr.columns, rotation=90)
plt.yticks(range(len(corr.columns)), corr.columns)

plt.title("Correlation Matrix")
plt.tight_layout()
plt.show()

print("\nCorrelation with Yield")
print(corr["Yield_Tonnes_Ha"].sort_values(ascending=False))

# -------------------------------
# Scatter Plot
# -------------------------------
plt.figure(figsize=(6,4))
plt.scatter(df["Rainfall_mm"], df["Yield_Tonnes_Ha"], alpha=0.5)
plt.xlabel("Rainfall (mm)")
plt.ylabel("Yield (Tonnes/Ha)")
plt.title("Rainfall vs Yield")
plt.grid(True)
plt.show()

# -------------------------------
# Seasonal Statistics
# -------------------------------
stats = df.groupby("Season")[["Yield_Tonnes_Ha","Profit_INR"]].agg(["mean","std"])

print("\nSeasonal Statistics")
print(stats)

# -------------------------------
# Final Insights
# -------------------------------
print("\n"+"="*60)
print("FINAL PROJECT INSIGHTS")
print("="*60)

print("Highest Yield Season      :", yield_season.idxmax())
print("Lowest Yield Season       :", yield_season.idxmin())

print("Highest Profit Season     :", profit.idxmax())
print("Lowest Profit Season      :", profit.idxmin())

print("Highest Rainfall Season   :", rain.idxmax())
print("Highest Temperature Season:", temp.idxmax())

print("Most Water Efficient      :", eff.idxmax())

print("\nTop 5 Crops by Production")
print(crop.head())

print("\nTop 5 States by Profit")
print(state.head())

print("\nProject Completed Successfully!")
```
